```python
# FIFA World Cup 2026 - Final Insights 

print("Total Matches :", len(data))
print("Total Teams :", len(set(data["HOME_TEAM"]).union(set(data["AWAY_TEAM"]))))
print("Tournament Rounds :", data["ROUND"].nunique())
print("Host Venues :", data["VENUE"].nunique())
```

```python
print("Average Goals per Match :", round(data["TOTAL_GOALS"].mean(),2))
print("Highest Goals in a Match :", data["TOTAL_GOALS"].max())
print("Total Goals :", data["TOTAL_GOALS"].sum())
```

```python
print("Average Attendance :", round(data["ATTENDANCE"].mean(),2))
print("Highest Attendance :", data["ATTENDANCE"].max())
print("Lowest Attendance :", data["ATTENDANCE"].min())
```

```python
print("Average Home Possession :", round(data["HOME_POSSESSION"].mean(),2))
print("Average Away Possession :", round(data["AWAY_POSSESSION"].mean(),2))
```

```python
print("Average Home Shots :", round(data["HOME_TOTAL_SHOTS"].mean(),2))
print("Average Away Shots :", round(data["AWAY_TOTAL_SHOTS"].mean(),2))

print("Average Home Shots on Target :", round(data["HOME_SOT"].mean(),2))
print("Average Away Shots on Target :", round(data["AWAY_SOT"].mean(),2))
```

```python
print("Average Home Fouls :", round(data["HOME_FOULS"].mean(),2))
print("Average Away Fouls :", round(data["AWAY_FOULS"].mean(),2))

print("Average Home Yellow Cards :", round(data["HOME_CARDS_YELLOW"].mean(),2))
print("Average Away Yellow Cards :", round(data["AWAY_CARDS_YELLOW"].mean(),2))
```
