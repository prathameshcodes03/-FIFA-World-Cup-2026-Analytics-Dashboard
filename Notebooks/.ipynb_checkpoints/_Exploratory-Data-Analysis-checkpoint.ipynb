```python

match_results = data["MATCH_RESULT"].value_counts()

print(match_results)
```

```python
plt.figure(figsize=(8,5))

plt.bar(match_results.index, match_results.values,color = 'orange')

plt.title("Match Result Distribution")

plt.xlabel("Match Result")

plt.ylabel("Number of Matches")
plt.show()
```

```python
# Sort matches by total goals scored
highest_scoring_matches = data.sort_values(
    by="TOTAL_GOALS",
    ascending=False
)

# Display Top 10 highest scoring matches
highest_scoring_matches[
    ["HOME_TEAM", "AWAY_TEAM", "HOME_SCORE", "AWAY_SCORE", "TOTAL_GOALS"]
].head(10)
```

```python
top10 = highest_scoring_matches.head(10)

match_names = top10["HOME_TEAM"] + " vs " + top10["AWAY_TEAM"]

plt.figure(figsize=(10,6))

plt.barh(match_names, top10["TOTAL_GOALS"] , color = 'green')

plt.title("Top 10 Highest Scoring Matches")

plt.xlabel("Total Goals")

plt.ylabel("Match")

plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.hist(x=data['TOTAL_GOALS'],color = 'darkblue')
plt.title("Goals Distribution")
plt.xlabel("Total Goals per Match")
plt.ylabel("Number of Matches")
plt.show()
```

```python
# Highest Scoring Match
highest_match = data.loc[data["TOTAL_GOALS"].idxmax()]

print("Highest Scoring Match")
print("----------------------")
print(f"{highest_match['HOME_TEAM']} vs {highest_match['AWAY_TEAM']}")
print(f"Score: {highest_match['HOME_SCORE']} - {highest_match['AWAY_SCORE']}")
print(f"Total Goals: {highest_match['TOTAL_GOALS']}")
```

```python
# Lowest Scoring Match
lowest_match = data.loc[data["TOTAL_GOALS"].idxmin()]

print("Lowest Scoring Match")
print("---------------------")
print(f"{lowest_match['HOME_TEAM']} vs {lowest_match['AWAY_TEAM']}")
print(f"Score: {lowest_match['HOME_SCORE']} - {lowest_match['AWAY_SCORE']}")
print(f"Total Goals: {lowest_match['TOTAL_GOALS']}")
```

```python
# Home wins
home_wins = data[data["MATCH_RESULT"] == "Home Win"]["HOME_TEAM"].value_counts()

# Away wins
away_wins = data[data["MATCH_RESULT"] == "Away Win"]["AWAY_TEAM"].value_counts()

# Total wins
total_wins = home_wins.add(away_wins, fill_value=0).astype(int)

# Sort descending
total_wins = total_wins.sort_values(ascending=False)

total_wins
```

```python
plt.figure(figsize=(10,6))

plt.bar(total_wins.index, total_wins.values,color = 'Teal')

plt.title("Teams with Most Wins")
plt.xlabel("Team")
plt.ylabel("Number of Wins")

plt.xticks(rotation=90)

plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()
```

```python
# Goals scored at home
home_goals = data.groupby("HOME_TEAM")["HOME_SCORE"].sum()

# Goals scored away
away_goals = data.groupby("AWAY_TEAM")["AWAY_SCORE"].sum()

# Total goals
total_goals = home_goals.add(away_goals, fill_value=0).astype(int)

total_goals = total_goals.sort_values(ascending=False)

total_goals
```

```python
plt.figure(figsize=(10,6))

plt.bar(total_goals.head(10).index, total_goals.head(10).values,color = 'purple')

plt.title("Top 10 Highest Scoring Teams")
plt.xlabel("Team")
plt.ylabel("Goals Scored")

plt.xticks(rotation=45)

plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
# Goals conceded at home
home_conceded = data.groupby("HOME_TEAM")["AWAY_SCORE"].sum()

# Goals conceded away
away_conceded = data.groupby("AWAY_TEAM")["HOME_SCORE"].sum()

# Total goals conceded
goals_conceded = home_conceded.add(away_conceded, fill_value=0).astype(int)

goals_conceded = goals_conceded.sort_values(ascending=False)

goals_conceded
```

```python
plt.figure(figsize=(10,6))

plt.bar(goals_conceded.head(10).index, goals_conceded.head(10).values,color = 'gray')

plt.title("Top 10 Teams Conceding Most Goals")
plt.xlabel("Team")
plt.ylabel("Goals Conceded")

plt.xticks(rotation=45)

plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()
```

```python
team_summary = pd.DataFrame({
    "Wins": total_wins,
    "Goals Scored": total_goals,
    "Goals Conceded": goals_conceded
}).fillna(0).astype(int)

team_summary = team_summary.sort_values("Wins", ascending=False)

team_summary
```

```python
numerical_columns = data.select_dtypes(include=["int64", "float64"])

numerical_columns.head()
```

```python
correlation_matrix = numerical_columns.corr()

correlation_matrix
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14,10))

plt.imshow(correlation_matrix, cmap="coolwarm", aspect="auto")

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.show()
```

```python
correlation_matrix.unstack().sort_values(ascending=False).drop_duplicates().head(15)
```

```python
correlation_matrix.unstack().sort_values().head(15)
```

```python
goal_correlation = correlation_matrix["TOTAL_GOALS"].sort_values(ascending=False)

goal_correlation
```

```python
plt.figure(figsize=(8,6))

goal_correlation.drop("TOTAL_GOALS").head(10).plot(kind="bar")

plt.title("Top Features Correlated with Total Goals")

plt.ylabel("Correlation Coefficient")

plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()
```
